Jupyter Lab is an integrated development environment (IDE) which runs on an interactive computing framework called `IPython`.

## Variables

A variable can basically represent anything! It is a convenient way of naming (or labeling) an object and storing it in your computer's memory during a Python session. This is so that you can access it later and manipulate it, if needed.

```python
x = 365
```

Since a variable is just referring to an object, it will have the attributes and methods of the object.
An attribute is a value associated with that object, while a method is a function specific to that object. More on functions later...

```python
x.imag
```

```python
# you must include the parentheses to use a method, otherwise you will just get the object
x.is_integer()
```

You can overwrite a variable by assigning a new value to it. In other words, you can change a variable. See why it's called a variable?

```python
x = "apple"
```

```python
x
```

```python
# now the variable has different methods
x.capitalize()
```

Variable naming is important!
At some point, you will have a lot of variables and it can be hard to keep track of what they mean if they are named `x`, `a`, `b`, `c`, etc.

In Python variable names...
* Must start with a letter or `_`, they cannot start with a number
* Can contain only alphanumeric characters or `_` (i.e. `A$APRocky` is not a valid variable name)
* Are case-sensitive (`JupyterLab` and `jupyterlab` are two different variables)

If you use an invalid variable name, you will get a `SyntaxError`.
**Syntax** is the set of grammatical rules which govern a specific programming language.
If you do not use the correct syntax, the computer will not know what you are trying to do and get confused!
An error message is a way of the computer communicating with you, asking for help.
