```python
import cmath

def clean_complex(c):
    """Rounds values near zero caused by floating-point inaccuracy."""
    real = round(c.real, 4) if abs(c.real) > 1e-12 else 0.0
    imag = round(c.imag, 4) if abs(c.imag) > 1e-12 else 0.0
    sign = "+" if imag >= 0 else "-"
    return f"{real} {sign} {abs(imag)}j"

# Define complex pairs z and w
problems = [
    ("Part 1", 
     3 + 0j, 
     1 - 2j),
    
    ("Part 2", 
     2 + 2j, 
     cmath.exp(1j * cmath.pi / 3)),
    
    ("Part 3", 
     2 * cmath.exp(-1j * cmath.pi / 2), 
     2 * cmath.exp(1j * 2 * cmath.pi / 3))
]

# Compute and display results
for title, z, w in problems:
    z_sum = z + w
    z_prod = z * w
    
    print(f"=== {title} ===")
    print(f"  z     = {clean_complex(z)}")
    print(f"  w     = {clean_complex(w)}")
    print(f"  z + w = {clean_complex(z_sum)}")
    print(f"  z * w = {clean_complex(z_prod)}\n")
```
