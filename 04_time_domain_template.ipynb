### Half-Wave Rectification

Write a function called `half_rectify` which performs half-wave rectification and takes the following inputs:
* `in_file`: the name of the WAV file you want to rectify
* `out_file`: the name of rectified WAV file to save to your disk

In the function, you will load the `in_file`, rectify it, and then save it as the `out_file`. For this function, I want you to use a `for` loop with conditionals (i.e., `if`, `elif`, `else`, etc.)

```python
import soundfile as sf
def half_rectify(in_file, out_file):
    """Half-wave rectification.

    Args:
        in_file (str): path to input WAV file
        out_file (str): path to save rectified WAV file
    """
    # YOUR CODE HERE
    y, sr = sf.read(in_file)
    for i in range(len(y)):
        if y[i] < 0:
            y[i] = 0
        else:
            y[i] = y[i]
    sf.write(out_file, y, sr)
```

Now, call your function on `sawtooth.wav`. Play back the rectified audio.

```python
half_rectify('Audio/sawtooth.wav', 'Audio/sawtooth_half.wav')
```
