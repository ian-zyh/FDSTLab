## Data Structures

So far, we have been working with objects that hold a single value. What if we want to collect multiple values together?

There are several different types of data containers in base Python. We will focus on the most important ones.

The following data structures can hold data of any type.
* Lists
* Tuples
* Dictionaries (not used in this class)

In order to select a piece of data from a structure, we need to _index_ it.
Python is a zero-indexed programming language, meaning that the first item
in a structure has an index of **`0`**.

### Tuples

A tuple is a sequence of objects separated by commas, enclosed in parentheses `()`.

* Ordered
* Allow duplicate values
* Fixed length
* Cannot be changed (immutable)

````python
lotto_num = (57, 12, 49, 36, 13, 40)
````

````python
# positive indexing
lotto_num[1]
````

````python
# negative indexing
lotto_num[-1]
````

````python
# tuples are immutable so if you try to change one of the values you will get an error!
lotto_num[4] = 19
````

### Lists

A list is a sequence of objects separated by commas, enclosed in square brackets `[]`.

* Ordered
* Allow duplicate values
* Variable length
* Can be changed (mutable)

````python
fruits = ['apple', 'banana', 'pear', 'grape', 'cherry', 'orange']
````

````python
# we can index lists similarly to tuples
fruits[0]
````

````python
# what happens if we use an index that is too high?
fruits[11]
````

````python
# we can find the number of items in a list
len(fruits)
````

Selecting a portion of a list (and some other data structures) is known as _slicing_.
It creates a new list with elements from the _start_ (inclusive) and _stop_ (exclusive) values.

````python
# we can select multiple items from a list
fruits[2:5]
````

````python
# we can add to the list
fruits.append('kiwi')
fruits
````

````python
# we can modify the list
fruits[0] = 'honeycrisp apple'
fruits
````

````python
# we can also combine lists
veggies = ['onion', 'carrot', 'spinach', 'broccoli']
produce = fruits + veggies
produce
````

You can even make a list of lists. This is called a _nested_ list.

````python
matrix = [[9, 1, 4], [5, 2, 8], [2, 8, 6]]
````

````python
# the first item in the list is itself a list
matrix[0]
````

````python
# we can index twice to get a specific value
matrix[0][1]
````

````python
# we can still change one of the numbers
matrix[1][2] = 3
matrix
````

## User-Defined Functions

A _function_ is very similar to the kind of function you learned about in your early math classes.
It takes an input, performs some sort of operation, and gives you an output.

The annoying $y=mx+b$ equation is a function.
It takes a value for $x$, multiplies it by the slope $m$, and adds the $y$-intercept $b$.
This gives you a value for $y$.

We can implement this in Python!

````python
# y = 2x + 1
m = 2
b = 1

x = 4
````

````python

y = m * x + b
print(f"If x = {x}, then y = {y}") 
````

Now, let's say that I want to find out what $y$ is if $x = -1$.

I have re-assign `x` and re-write the equation. This becomes even more annoying if I want to change the slope or intercept!

````python

x = -1
y = m * x + b
print(f"If x = {x}, then y = {y}") 
````

To avoid copying and pasting lines of code over and over again, we can define our own _function_!

There is a very specific structure that you need to follow.

```
def function_name(arg1, arg2, ...):
    body of statements
    return
```
If there is one quirk about Python, it is that **indentation matters**. This white space helps to structure the code and make it more easily readable. If you do not properly indent, you will get an `IndentationError`.

An indentation is equivalent to 4 spaces or one `Tab`.

````python
# I can turn my linear equation into a function
def line(x, m, b):
    y = m * x + b
    return y
````

````python
# I can call my function after defining it
line(-1, 2, 1)
````

````python
# I can easily change my parameters
line(-1, 3, 0)
````

What if I want my function to use some default values for certain arguments or parameters? 

Then, you can set the default value in the definition of the function. This parameter then becomes _optional_. When you call the function, if you do not explicitly provide a parameter, it will use the default one.

````python
def line(x, m=1, b=0):
    y = m * x + b
    return y
````

````python
# y = x
line(1)
````

````python
# y = x + 5
line(1, 5)
````

````python
# y = 0.5x
# if you do not use the name of the argument, Python will guess what you are referring to by position
line(1, 0.5)
````

Sometimes, your function may not have any input and/or output.
That's ok! You still need to use the parentheses `()` to call it.

````python
def hello_world():
    print("hello world!")
    print("that is all there is to say")
````

````python
hello_world()
````

````python
# if you do not have the () it just returns the function as an object
line(1, 0.5)
````

It is extremely important to understand that variables you define in your function, only exist _within_ your function.

If you try to access this variable outside of your function definition, you will either get an error (`NameError`) or be referring to the wrong variable!

These variables definied in your function are _local_ to your function, while the ones you define outside in your environment are _global_.

````python
# this is global
borough = "Manhattan"
````

## Conditionals

This is where programming starts to become powerful.
Using conditional statements, you can tell the computer what to do depending on whether a certain statement is true or not (remember booleans?).
```
if condition:
    statement
```
There are also different conditions and boolean operators which are useful in `if` statements.
* `a == b`: check for equality between `a` and `b`
* `a < b`: check if `a` is less than `b`
* `a <= b`: check if `a` is less than or equal to `b`
* `a > b`: check if `a` is greater than `b`
* `a >= b`: check if `a` is greater than or equal to `b`
* `a != b`: check if `a` is *not* equal to `b`
* `c and d`: both conditions must be `True`
* `c or d`: at least one condition must be `True`
* `not c`: the condition must be not `True` (so `False`)
* `f in e`: check if `f` is contained in `e` (usually a list)

````python
# check if a number is negative

x = -45

if x < 0:
    print(f"{x} is negative")
else:
    print(f"{x} is non-negative")
````

If we want the computer to something different when the condition is `False`, we can add the `else` clause.
```
if condition:
    statement1
else:
    statement2
```

````python
x = 3

if x < 0:
    print(f"{x} is a negative number")
else:
    print(f"{x} is a non-negative number")
````

If we have multiple conditions we want to check, then we can use `elif`
```
if condition1:
    statement1
elif condition2:
    statement2
else:
    statement3
```

````python
x = 29

if x < 0:
    print(f"{x} is a negative number")
elif x > 0:
    print(f"{x} is a positive number")
else:
    print(f"{x} is neither a positive nor negative number")
````

## Loops

Loops allow you to run certain lines of code multiple times.
There are two types of loops in Python: `while` loops and `for` loops.

### `while` Loops

In a `while` loop, the statements in the block will be executed repeatedly as long at the condition is `True`.
```
while condition:
    statement(s)
```
With these kinds of loops, you need to be careful to make sure that eventually the loop is broken.
This is usually done by making changes within the loop so that eventually the condition will be `False`.
If you do not do this, you will get an _infinite_ loop!

````python
i = 0

while i < 10:
    print(i)
    i += 1
````

### `for` Loops

In a `for` loop, you _iterate_ over a finite sequence (e.g., list, ranges, etc.).
The statement(s) in the loop is executed once for each item in the sequence.
```
for value in sequence:
    statement(s)
```

````python
for i in range(10):
    print(i)
````
