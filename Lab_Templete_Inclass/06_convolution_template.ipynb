The convolution of two sequences $x[n]$ of length $N$ and $h[k]$ of length $K$ is defined by:

$y[n] = \sum_{k=0}^{K-1} h[k]\cdot x[n-k]$

## `np.convolve`

NumPy has an predefined convolution function [`np.convolve`](https://numpy.org/doc/2.0/reference/generated/numpy.convolve.html) that lets you perform 1D convolution between two signals (`np.ndarray`) and specify the padding mode.

Let's use it to see if we get the same results as our in-class example.

$x = [5, 6, 7, 8, 9]$

$h = [1, 0, -1]$

What is the default mode?

```python
x = np.arange(5, 10)
h = np.array([1, 0, -1])
```

```python
np.convolve(x, h)
```

Try convolution on the same signal and filter with the other two modes.

```python
np.convolve(x, h, mode='same')
```

## Direct Convolution

Write a function called `convolve` which takes in a signal `x` and a filter `h` and directly convolves them in the "full" mode. You must compute the convolution manually (i.e., you cannot use `np.convolve`).

```python
def convolve(x, h):
    '''Convolve two 1D signals together in "full" mode.

    Parameters
    ----------
    x : (np.ndarray)
        The input signal
        
    h : (np.ndarray)
        The filter signal
        
    Returns
    -------
    y : (np.ndarray)
        The convolved signal
    '''

    N = len(x)
    M = len(h)
    y = np.zeros(N + M - 1)
    
    for k in range(N):
        for m in range(M):
            y[k + m] += x[k] * h[m]
    return y
```

Test your function on the signal and filter below. Check that your output is correct using `np.convolve`.

```python
x = np.array([1, 1, 4, 2])
h = np.array([1, 0, 2, 0, 1])
```

```python
# call your function here
y = convolve(x, h)
print(y)
```

```python
# call np.convolve here
y = np.convolve(x, h, mode='full')
print(y)
```

```python
# checks that the arrays are equal
assert sum(~np.isclose(convolve(x, h), np.convolve(x, h))) == 0
```
